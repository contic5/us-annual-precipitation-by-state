```python
import pandas as pd
```

```python
drought_df=pd.read_excel("US-Annual-Precipitation-By-State.xlsx",sheet_name="Palmer_Drought_Severity_Index")
print(drought_df.head())
```

```python
def pivot_by_decade(df,area_type):
    decade_df=df.pivot_table(
        index="State",
        columns="Decade",
        values="Avg",
        aggfunc="mean"
    )

    decade_df=decade_df.reset_index()
    print(decade_df.columns)
    decade_df=decade_df.sort_values(by=2020,ascending=True)
    decade_df.to_excel(f"output/palmer_drought_by_decade/{area_type}.xlsx",index=False)
```

```python
drought_state_df=drought_df[~pd.isna(drought_df["Region"])]
drought_region_df=drought_df[pd.isna(drought_df["Region"])]

pivot_by_decade(drought_state_df,"state")
pivot_by_decade(drought_region_df,"region")
```
