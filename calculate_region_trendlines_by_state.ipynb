```python
from scipy import stats
import pandas as pd
```

```python
drought_df=pd.read_excel("US-Annual-Precipitation-By-State.xlsx",sheet_name="Palmer_Drought_Severity_Index")
drought_df=drought_df[~pd.isna(drought_df["Region"])]
unique_regions=list(drought_df["Region"].unique())
```

```python
region_drought_list=[]

row=1
for region in unique_regions:
    drought_df_filtered=drought_df[drought_df["Region"]==region]
    drought_df_filtered = drought_df_filtered.groupby(['Decade'], as_index=False)['Avg'].mean()
    drought_df_filtered["Year_Distance"]=drought_df_filtered["Decade"]-drought_df_filtered["Decade"].min()
    
    x=drought_df_filtered["Year_Distance"]
    y=drought_df_filtered["Avg"]

    #r_value is strength of correlation and p_value is change that value is random
    slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
    trend_y = slope * x + intercept
    region_drought_dict={"ID":row,"Region":region,"Slope":slope,"Intercept":intercept,"Correlation":r_value,"Significance":p_value,"Standard_Error":std_err,"Datapoints":len(drought_df_filtered)}
    region_drought_list.append(region_drought_dict)
    row+=1
```

```python
region_drought_df=pd.DataFrame.from_dict(region_drought_list)
region_drought_df=region_drought_df.sort_values(by="Significance",ascending=False)
region_drought_df.to_excel("US-Palmer-Drought-Severity-By-Region.xlsx")
```
