```python
import pandas as pd
import matplotlib.pyplot as plt

from collections import defaultdict
import math
```

```python
drought_df=pd.read_excel("US-Annual-Precipitation-By-State.xlsx",sheet_name="Palmer_Drought_Severity_Index")
print(drought_df.head())
```

```python
def blank():
    return 0

drought_group_maxes={"Extreme Drought":-4,"Severe Drought":-3,"Moderate Drought":-2,"Near Normal":1.99,"Unusually Moist Spell":3,"Very Moist Spell":4,"Extremely Moist":10
          }
def get_counts_by_time(df,time_column):
    df = df.groupby(['Decade','State'], as_index=False)['Summer_Avg'].mean()

    decade_counts=[]
    for i in range(df["Decade"].min(),df["Decade"].max()+9,10):
        decade_counts.append(defaultdict(blank))
        decade_counts[-1]["Decade"]=i

    for index,row in df.iterrows():
        decade_index=(row["Decade"]-df["Decade"].min())/10
        decade_index=decade_index.astype(int)
        avg_val=row["Summer_Avg"]

        for group in drought_group_maxes:
            if avg_val<drought_group_maxes[group]:
                decade_counts[decade_index][group]+=1
                break

    deacade_df=pd.DataFrame.from_dict(decade_counts)
    print(deacade_df.head())
    deacade_df = deacade_df.fillna(0)
    deacade_df.to_excel("State Drought Levels by Decade.xlsx")
    
```

```python
drought_state_df=drought_df[~pd.isna(drought_df["Region"])]
drought_state_df["Summer_Avg"]=drought_state_df[['Jun','Jul','Aug']].mean(axis=1)
#drought_region_df=drought_df[pd.isna(drought_df["Region"])]
print(drought_state_df.head())
get_counts_by_time(drought_state_df,"Decade")
```
