```python
from scipy import stats
import pandas as pd
```

```python
drought_df=pd.read_excel("US-Annual-Precipitation-By-State.xlsx",sheet_name="Palmer_Drought_Severity_Index")

```

```python
def create_data_frame(df,area_column:str,time_column:str,averaging:bool):
    if area_column=="State":
        df=df[~pd.isna(df["Region"])]
    elif area_column=="Region":
         df=df[pd.isna(df["Region"])]

    unique_areas=list(df["State"].unique())

    area_drought_list=[]
    row=1
    for region in unique_areas:
        df_filtered=df[df["State"]==region]
        if averaging:
            df_filtered = df_filtered.groupby([time_column], as_index=False)['Avg'].mean()
        
        x=df_filtered[time_column]-df_filtered[time_column].min()
        print(x.describe())
        y=df_filtered["Avg"]

        #r_value is strength of correlation and p_value is change that value is random
        slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
        trend_y = slope * (df_filtered[time_column].max()-df_filtered[time_column].min()) + intercept
        region_drought_dict={"ID":row,area_column.title():region,"Slope":slope,"Intercept":intercept,"Correlation":r_value,"(R) Significance":p_value,"(P) Standard_Error":std_err,"Datapoints":len(df_filtered),"Trend Y":trend_y}
        area_drought_list.append(region_drought_dict)
        row+=1

    area_drought_df=pd.DataFrame.from_dict(area_drought_list)
    area_drought_df=area_drought_df[area_drought_df["Datapoints"]>0]
    area_drought_df=area_drought_df.sort_values(by="Slope",ascending=True)

    file_name=f"{time_column.title()}.xlsx"
    if averaging:
        file_name=file_name.replace(".xlsx","-Averaged.xlsx")
    area_drought_df.to_excel(f"output/palmer_drought_trendlines/{area_column.lower()}/{file_name}",index=False)
```

```python

averaging_options=[False]
time_columns=["Year","Half-Decade","Decade"]

for averaging_option in averaging_options:
    for target_column in time_columns:
        create_data_frame(drought_df,"Region",target_column,averaging_option)
        create_data_frame(drought_df,"State",target_column,averaging_option)
```
